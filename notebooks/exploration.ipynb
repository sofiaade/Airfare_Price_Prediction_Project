# Airfare Price Prediction — Exploratory Analysis
Uses the Flight Price Prediction dataset (Kaggle).

**Goal**: Predict domestic Indian flight prices using features like airline, route, stops, duration, and days until departure.

```python
import sys
sys.path.insert(0, '../src')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

from preprocess import download_dataset, get_feature_dataframe

csv_path = download_dataset('../data/')
df = get_feature_dataframe(csv_path)
print(f'Shape: {df.shape}')
df.head()
```

## 1. Price Distribution

```python
fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(df['price'], bins=60, color='#2E86AB', edgecolor='white', alpha=0.85)
ax.axvline(df['price'].median(), color='orange', linestyle='--', label=f"Median: ₹{df['price'].median():,.0f}")
ax.set_title('Distribution of Airfare Prices')
ax.set_xlabel('Price (INR)')
ax.legend()
plt.tight_layout()
```

## 2. Price by Class

```python
df.groupby('class')['price'].describe().round(0)
```

## 3. Correlation with numeric features

```python
numeric = df[['duration', 'days_left', 'stops', 'price']]
print(numeric.corr()['price'].sort_values(ascending=False))
```

## 4. Train models

```python
from train import train_all
results = train_all(data_dir='../data/', models_dir='../models/')

import pandas as pd
pd.DataFrame(results)[['model','mae','rmse','r2']].sort_values('r2', ascending=False)
```

## 5. Make a prediction

```python
from predict import AirfarePricePredictor
predictor = AirfarePricePredictor(models_dir='../models/')

price = predictor.predict_single(
    airline='IndiGo',
    source_city='Delhi',
    destination_city='Mumbai',
    flight_class='Economy',
    stops='zero',
    duration=2.5,
    days_left=10,
    departure_time='Morning',
    arrival_time='Afternoon',
)
print(f'Predicted price: ₹{price:,.0f}')
```
